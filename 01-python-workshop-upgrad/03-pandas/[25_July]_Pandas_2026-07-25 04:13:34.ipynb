```python
# Part 1 :
import pandas as pd

sales = pd.DataFrame({
    "OrderID": [101,102,103,104,105,106,107,108],
    "Customer": ["Sankara","Priya","Rahul","Sneha","Priya","Rahul","Sankara","Neha"],
    "City": ["Delhi","Mumbai","Delhi","Pune","Mumbai","Delhi","Delhi","Pune"],
    "Category": ["Laptop","Phone","Phone","Laptop","Tablet","Laptop","Tablet","Phone"],
    "Amount": [80000,50000,35000,90000,25000,85000,30000,45000]
})

sales
```

```python
# Question 1
# Find the total sales (Amount) for each city.
# Question 2
# Find the average order amount for each product category.
# Question 3 (Multiple Aggregations)
# For each city, find:
# Total Sales
# Average Sales
# Maximum Sales
# Number of Orders
```

```python
total_sales_by_city = sales.groupby('City')['Amount'].sum()
print(total_sales_by_city)
```

```python
avg_order_amount_by_category = sales.groupby('Category')['Amount'].mean()
print(avg_order_amount_by_category)
# For each city, find:
# Total Sales
# Average Sales
# Maximum Sales
# Number of Orders
sales_by_city = sales.groupby('City').agg({
    'Amount': ['sum', 'mean', 'max', 'count']
})
print(sales_by_city)
```

```python
customers = pd.DataFrame({
    "CustomerID": [1,2,3,4],
    "CustomerName": ["Sankara","Priya","Rahul","Sneha"],
    "City": ["Delhi","Mumbai","Delhi","Pune"]
})

orders = pd.DataFrame({
    "OrderID": [101,102,103,104,105],
    "CustomerID": [1,2,3,5,2],
    "Amount": [5000,7000,9000,6000,8000]
})
```

```python
# Question 1
# Perform a Left Join between customers and orders. Which customers have not placed any order?
customers_left_join = pd.merge(customers, orders, on='CustomerID', how='left')
customers_left_join
```

```python
orders
```

```python
# Question 2
# Perform an Outer Join. Which order belongs to a customer who is not present in the customer table?
#customers_outer_join = pd.merge(customers, orders, on='CustomerID', how='right')
customers_outer_join = customers.merge(orders, on='CustomerID', how='right')
customers_outer_join
```

```python
# Question 3
# Perform a Right Join. Which customers have not placed any order?
customers_right_join = pd.merge(customers, orders, on='CustomerID', how='outer')
customers_right_join
```
