You are given a total number of seconds. Convert it to hours, minutes, and seconds and 
print it in the format hh:mm:ss.



Sample Input:

3661



Sample Output:

01:01:01

```python
import datetime 
# Taking input
total_seconds = int(input())

# Write your code here

hours = total_seconds // 3600
minutes = (total_seconds % 3600) // 60
seconds = total_seconds % 60
output = f"{hours}:{minutes}:{seconds}"
output = datetime.datetime.strptime(output, "%H:%M:%S").strftime("%H:%M:%S")
# = (((total_seconds)/60)/60)
# Print the output
print(output)
```

Instructions

A vehicle travels two different segments of a journey:



1. It covers the first segment at speed s1 for time t1

2. Then it covers the second segment at speed s2 for time t2



Calculate and print the overall average speed.

Note: The average speed should be a whole number



Sample Input:

60

2

90

1



Sample Output:

70

```python
# Taking input
s1 = float(input())
t1 = float(input())
s2 = float(input())
t2 = float(input())

# Write your code here
d2=(s2*t2)
d1=(s1*t1)
avg_speed = int((d1+d2)/(t1+t2))
# Print the output
print(avg_speed)
```
